```python
import pandas as pd

df_train = pd.read_csv("train_data.csv")
df_test = pd.read_csv("test_data.csv")

df_train.head()
```

```python
subtask_1 = len(df_train)
subtask_2 = len(df_train[df_train.Gender == 'male'])
subtask_3 = df_train.Duration.mean()
subtask_4 = len(df_train[df_train.Age >= 75])

subtasks = pd.DataFrame({
    "subtaskID": [1, 2, 3, 4],
    "datapointID": 1,
    "answer": [subtask_1, subtask_2, subtask_3, subtask_4]
})

subtasks
```

```python
df_train.Gender = df_train.Gender.map({"male": 0, "female": 1})
df_test.Gender = df_test.Gender.map({"male": 0, "female": 1})

df_test.Gender = df_test.Gender.fillna(df_test.Gender.mean())

features = ['Age', 'Height', 'Weight', 'Duration', 'Heart_Rate', 'Body_Temp', 'Gender']
target = 'Calories'

X = df_train[features]
y = df_train[target]

X_final = df_test[df_test.Subtask == 5]
X_final  = X_final[features]

X.head()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)
y_final = model.predict(X_final)
```

```python
subtask_5 = pd.DataFrame({
    "subtaskID": 5,
    "datapointID": df_test[df_test.Subtask == 5].User_ID,
    "answer": y_final
})

subtask_5.head()
```

```python
submission = pd.concat([subtasks, subtask_5])
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

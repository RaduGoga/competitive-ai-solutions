```python
# Importăm biblioteca pandas pentru a manipula datele
import pandas as pd 

df_train = pd.read_csv('train_data.csv')
df_test = pd.read_csv('test_data.csv')
df_train.head()
```

```python
# Adăugăm coloane noi în setul de date de test pentru fiecare subtask
df_test['Subtask1'] = 0.0  # Subtask1 va avea un rezultat de tip float
df_test['Subtask2'] = False  # Subtask2 va avea un rezultat de tip boolean
df_test['Subtask3'] = 0  # Subtask3 va avea un rezultat de tip întreg
df_test['Subtask4'] = 0  # Subtask4 va avea un rezultat de tip întreg
df_test['Subtask5'] = 0.0  # Subtask5 va avea un rezultat de tip float (aici vom adăuga predicțiile)

meanie = df_train.Hours_Studied.mean()
scores = df_train.Previous_Scores
def get_bigger_scores(val):
    cnt = 0
    for score in scores:
        if score >= val:
            cnt += 1
    return cnt
values = df_train.Motivation_Level.value_counts()

def subtasks(row):
    #subtask1
    row.Subtask1 = abs(row.Hours_Studied - meanie)
    #subtask2
    if row.Sleep_Hours < 7:
        row.Subtask2 = True
    else:
        row.Subtask2 = False
    #subtask3
    row.Subtask3 = get_bigger_scores(row.Previous_Scores)
    #subtask4
    row.Subtask4 = values[row.Motivation_Level]
    return row

df_test = df_test.apply(subtasks, axis='columns')
```

```python
#preprocessing ggggg
# AM FACUT LABEL ENCDODER DE MANA :(((((

df_train.Teacher_Quality = df_train.Teacher_Quality.fillna('Medium')
df_train.Parental_Education_Level = df_train.Parental_Education_Level.fillna('High School')
df_train.Distance_from_Home = df_train.Distance_from_Home.fillna('Moderate')

df_train.Parental_Involvement = df_train.Parental_Involvement.map({"Low": 0, "Medium": 1, "High": 2})
df_train.Access_to_Resources = df_train.Access_to_Resources.map({"Low": 0, "Medium": 1, "High": 2})
df_train.Extracurricular_Activities = df_train.Extracurricular_Activities.map({"No": 0, "Yes": 1})
df_train.Motivation_Level = df_train.Motivation_Level.map({"Low": 0, "Medium": 1, "High": 2})
df_train.Internet_Access = df_train.Internet_Access.map({"No": 0, "Yes": 1})
df_train.Family_Income = df_train.Family_Income.map({"Low": 0, "Medium": 1, "High": 2})
df_train.Teacher_Quality = df_train.Teacher_Quality.map({"Low": 0, "Medium": 1, "High": 2})
df_train.School_Type = df_train.School_Type.map({"Public": 0, "Private": 1})
df_train.Peer_Influence = df_train.Peer_Influence.map({"Negative": 0, "Neutral": 1, "Positive": 2})
df_train.Learning_Disabilities = df_train.Learning_Disabilities.map({"No": 0, "Yes": 1})
df_train.Parental_Education_Level = df_train.Parental_Education_Level.map({"High School": 0, "College": 1, "Postgraduate": 2})
df_train.Distance_from_Home = df_train.Distance_from_Home.map({"Near": 0, "Moderate": 1, "Far": 2})

df_test.Teacher_Quality = df_test.Teacher_Quality.fillna('Medium')
df_test.Parental_Education_Level = df_test.Parental_Education_Level.fillna('High School')
df_test.Distance_from_Home = df_test.Distance_from_Home.fillna('Moderate')

df_test.Parental_Involvement = df_test.Parental_Involvement.map({"Low": 0, "Medium": 1, "High": 2})
df_test.Access_to_Resources = df_test.Access_to_Resources.map({"Low": 0, "Medium": 1, "High": 2})
df_test.Extracurricular_Activities = df_test.Extracurricular_Activities.map({"No": 0, "Yes": 1})
df_test.Motivation_Level = df_test.Motivation_Level.map({"Low": 0, "Medium": 1, "High": 2})
df_test.Internet_Access = df_test.Internet_Access.map({"No": 0, "Yes": 1})
df_test.Family_Income = df_test.Family_Income.map({"Low": 0, "Medium": 1, "High": 2})
df_test.Teacher_Quality = df_test.Teacher_Quality.map({"Low": 0, "Medium": 1, "High": 2})
df_test.School_Type = df_test.School_Type.map({"Public": 0, "Private": 1})
df_test.Peer_Influence = df_test.Peer_Influence.map({"Negative": 0, "Neutral": 1, "Positive": 2})
df_test.Learning_Disabilities = df_test.Learning_Disabilities.map({"No": 0, "Yes": 1})
df_test.Parental_Education_Level = df_test.Parental_Education_Level.map({"High School": 0, "College": 1, "Postgraduate": 2})
df_test.Distance_from_Home = df_test.Distance_from_Home.map({"Near": 0, "Moderate": 1, "Far": 2})

df_test.head()
```

```python
features = ['Hours_Studied', 'Attendance', 'Parental_Involvement', 'Access_to_Resources', 'Extracurricular_Activities', 'Sleep_Hours',
       'Previous_Scores', 'Motivation_Level', 'Internet_Access', 'Tutoring_Sessions', 'Family_Income', 'Teacher_Quality', 'School_Type',
       'Peer_Influence', 'Physical_Activity', 'Learning_Disabilities', 'Parental_Education_Level', 'Distance_from_Home']
target = 'Exam_Score'

X = df_train[features]
y = df_train[target]
X_final = df_test[features]

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_X = pd.DataFrame(scaler.fit_transform(X), columns=features)
scaled_X_final = pd.DataFrame(scaler.fit_transform(X_final), columns=features)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
from sklearn.linear_model import LinearRegression

X_train, X_valid, y_train, y_valid = train_test_split(scaled_X, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
y_preds = model.predict(X_valid)
mean_absolute_error(y_preds, y_valid)
```

```python
y_final = model.predict(scaled_X_final)

df_test.Subtask5 = y_final
```

```python
# Inițializăm o listă goală pentru a stoca rezultatele
result = []

# Iterăm prin fiecare rând al setului de date de test
for _, row in df_test.iterrows():
    # Iterăm prin subtasks (Subtask1 până la Subtask5)
    for subtask_id in range(1, 6):
        # Adăugăm un dicționar cu valorile corespunzătoare fiecărui subtask
        result.append({
            'subtaskID': subtask_id,  # ID-ul subtask-ului
            'datapointID': row['ID'],  # ID-ul datapoint-ului din rândul curent
            'answer': row[f'Subtask{subtask_id}']  # Răspunsul pentru subtask-ul curent
        })

# Creăm un DataFrame cu rezultatele obținute
df_output = pd.DataFrame(result)

# Salvăm rezultatele într-un fișier CSV pe care să-l putem apoi submite pe platformă
df_output.to_csv('submission.csv', index=False)
df_output
```

```python
import pandas as pd

df_train = pd.read_csv("train_data.csv")
df_test = pd.read_csv("test_data.csv")

df_test.head()
```

```python
#subtask 1

answers = []

for _, row in df_test.iterrows():
    BMI = row.weight_kg / ((row.height_cm/100)**2)
    if BMI > 25:
        answers.append(1)
    else:
        answers.append(0)

subtask_1 = pd.DataFrame({
    "subtaskID": 1,
    "datapointID": df_test.ID,
    "answer": answers
})

subtask_1.head()
```

```python
#preprocessing

df_train.smokes = df_train.smokes.map({"yes": 1, "no": 0, "0": 0, "1": 1})
df_train.gender = df_train.gender.map({"M": 0, "F": 1})

df_test.smokes = df_test.smokes.map({"yes": 1, "no": 0, "0": 0, "1": 1})
df_test.gender = df_test.gender.map({"M": 0, "F": 1})

df_train.sleep_hours = df_train.sleep_hours.fillna(df_train.sleep_hours.mean())
df_test.sleep_hours = df_test.sleep_hours.fillna(df_test.sleep_hours.mean())

df_test.head()
```

```python
#scaling

features = ['age', 'height_cm', 'weight_kg', 'heart_rate', 'blood_pressure', 'sleep_hours', 'nutrition_quality', 'activity_index', 'smokes', 'gender']
target = 'is_fit'

X = df_train[features]
y = df_train[target]

X_final = df_test[features]

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = pd.DataFrame(scaler.fit_transform(X), columns=features)
X_final = pd.DataFrame(scaler.transform(X_final), columns=features)

X.head()
```

```python
#splitting the data

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, stratify=y, random_state=42)
```

```python
#training and predicting

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import f1_score

model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)
y_preds = model.predict(X_test)
f1_score(y_preds, y_test)
```

```python
y_final = model.predict(X_final)
```

```python
subtask_2 = pd.DataFrame({
    "subtaskID": 2,
    "datapointID": df_test.ID,
    "answer": y_final
})

submission = pd.concat([subtask_1, subtask_2])

submission.to_csv("submission.csv")
```

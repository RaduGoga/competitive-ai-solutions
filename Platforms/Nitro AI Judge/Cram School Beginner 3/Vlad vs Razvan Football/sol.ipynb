```python
import pandas as pd

df_train = pd.read_csv("train_data.csv")
df_test = pd.read_csv("test_data.csv")

df_train.head()
```

```python
years = df_train.date.apply(lambda x : x[:4])
sb1 = years.mode()[0]

df_1 = pd.DataFrame({
    "subtaskID": 1,
    "datapointID": [0],
    "answer": sb1
})

df_1
```

```python
feats = [col for col in df_train.columns if df_train[col].dtype != 'str' and col != 'ID']
target = 'result'

X = df_train[feats]
y = df_train[target]

X_final = df_test[feats]

X.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

model = RandomForestClassifier()

#cross_val_score(model, X, y, cv=3, scoring='accuracy')
```

```python
model.fit(X, y)
y_preds = model.predict(X_final)
```

```python
df_2 = pd.DataFrame({
    "subtaskID": 2,
    "datapointID": df_test.ID,
    "answer": y_preds
})

df_2.head()
```

```python
submission = pd.concat([df_1, df_2])
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

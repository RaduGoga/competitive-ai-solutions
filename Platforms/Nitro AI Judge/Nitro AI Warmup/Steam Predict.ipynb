```python
import pandas as pd

df_train = pd.read_csv("train_data.csv")
df_test = pd.read_csv("test_data.csv")

df_train.head()
```

```python
#subtask1

def get_owners(value):
    minmax = value.split(' - ')
    media = (int(minmax[0]) + int(minmax[1])) / 2
    return media

df_test['Estimated owners'] = df_test['Estimated owners'].apply(get_owners)
df_train['Estimated owners'] = df_train['Estimated owners'].apply(get_owners)

subtask1 = pd.DataFrame({"subtaskID":1, "datapointID": df_test.AppID, "answer": df_test['Estimated owners']})

subtask1.head()
```

```python
#preprocess- uhhhhhhh
def get_year(value):
    return int(value.split(', ')[1])

df_test['Release date'] = df_test['Release date'].apply(get_year)
df_train['Release date'] = df_train['Release date'].apply(get_year)

features = ['Release date', 'Estimated owners', 'Metacritic score', 'Recommendations', 'Positive', 'Negative']
target = 'Price'

X = df_train[features]
y = df_train[target]
X_final = df_test[features]

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaled_X = scaler.fit_transform(X)
scaled_X = pd.DataFrame(scaled_X, columns=features)
scaled_X_final = scaler.transform(X_final)
scaled_X_final = pd.DataFrame(scaled_X_final, columns=features)

scaled_X
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.neighbors import KNeighborsRegressor

X_train, X_valid, y_train, y_valid = train_test_split(scaled_X, y, test_size = 0.8, random_state=42)

model = KNeighborsRegressor()
model.fit(X_train, y_train)
y_preds = model.predict(X_valid)
r2_score(y_preds, y_valid)
```

```python
final_preds = model.predict(scaled_X_final)
```

```python
subtask2 = pd.DataFrame({"subtaskID":2, "datapointID": df_test.AppID, "answer":final_preds})

submission = pd.concat([subtask1, subtask2])
submission.to_csv("sub.csv")
submission
```

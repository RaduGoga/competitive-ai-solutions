```python
import pandas as pd

df_train = pd.read_csv("train_data.csv")
df_test = pd.read_csv("test_data.csv")

df_train.head()
```

```python
df_train = df_train[(df_train.Class == 0) | (df_train.Class == 1)]

df_train.head()
```

```python
df_train.columns
```

```python
features = ['X', 'Y', 'Z']
target = 'Class'

X = df_train[features]
y = df_train[target]
X_final = df_test[features]

X.head()
```

```python
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

model = SVC(kernel='rbf', gamma=10, C=18)

cross_val_score(model, X, y, cv=5, scoring='f1_macro').mean()
```

```python
model.fit(X, y)
y_preds = model.predict(X_final)
```

```python
df_1 = pd.DataFrame({
    "subtaskID": 1,
    "datapointID": df_train.datapointID,
    "answer": df_train.Class
})

df_1.head()
```

```python
df_2 = pd.DataFrame({
    "subtaskID": 2,
    "datapointID": df_test.datapointID,
    "answer": y_preds
})

df_2.head()
```

```python
submission = pd.concat([df_1, df_2])
submission.answer = submission.answer.astype("int64")
submission.datapointID = submission.datapointID.astype("float64")
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

```python
import pandas as pd

df_train = pd.read_csv("train_data.csv")
df_test = pd.read_csv("test_data.csv")

df_train.head()
```

```python
def nutrient_score(row):
    return 0.4 * row.Nitrogen + 0.3 * row.Phosphorus + 0.3 * row.Potassium

subtask1 = df_test.apply(nutrient_score, axis=1)

df_1 = pd.DataFrame({
    "subtaskID": 1,
    "datapointID": df_test.ID,
    "answer": subtask1
})

df_1.head()
```

```python
def ph_class(val):
    if val < 6.0:
        return "Acid"
    elif val <= 7.5:
        return "Neutral"
    return "Alkaline"

subtask2 = df_test.pH.apply(ph_class)

df_2 = pd.DataFrame({
    "subtaskID": 2,
    "datapointID": df_test.ID,
    "answer": subtask2
})

df_2.head()
```

```python
mediana = df_train.Moisture.median()

def is_moisturer(val):
    if val > mediana:
        return 1
    return 0

subtask3 = df_test.Moisture.apply(is_moisturer)

df_3 = pd.DataFrame({
    "subtaskID": 3,
    "datapointID": df_test.ID,
    "answer": subtask3
})

df_3.head()
```

```python
value_dict = df_train.Soil_Type.value_counts()

subtask4 = df_test.Soil_Type.map(lambda x : value_dict[x])

df_4 = pd.DataFrame({
    "subtaskID": 4,
    "datapointID": df_test.ID,
    "answer": subtask4
})

df_4.head()
```

```python
#preprocesare simpla - imputing pt alea 3 coloane si un ordinal encoding pentru cele categorice

df_train.Soil_Type = df_train.Soil_Type.map({"Sandy": 0, "Loam": 1, "Clay": 2, "Silt": 3})
df_test.Soil_Type = df_test.Soil_Type.map({"Sandy": 0, "Loam": 1, "Clay": 2, "Silt": 3})

df_train.Region = df_train.Region.map({"Plain": 0, "Hill": 1, "Mountain": 2})
df_test.Region = df_test.Region.map({"Plain": 0, "Hill": 1, "Mountain": 2})

df_train.Irrigation = df_train.Irrigation.map({"None": 0, "Sprinkler": 1, "Drip": 2})
df_test.Irrigation = df_test.Irrigation.map({"None": 0, "Sprinkler": 1, "Drip": 2})

to_impute = ['Irrigation', 'Moisture', 'Organic_Matter']
for col in to_impute:
    df_train[col] = df_train[col].fillna(df_train[col].mean())
    df_test[col] = df_test[col].fillna(df_test[col].mean())

df_train.head()
```

```python
features = [col for col in df_test.columns if col != "ID"]
target = 'Suitability'

X = df_train[features]
y = df_train[target]
X_final = df_test[features]

X.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

model = RandomForestClassifier()

cross_val_score(model, X, y, cv=5, scoring='f1_macro').mean()
```

```python
model.fit(X, y)
y_preds = model.predict(X_final)
```

```python
df_preds = pd.DataFrame({
    "subtaskID": 5,
    "datapointID": df_test.ID,
    "answer": y_preds
})

df_preds.head()
```

```python
submission = pd.concat([df_1, df_2, df_3, df_4, df_preds])
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

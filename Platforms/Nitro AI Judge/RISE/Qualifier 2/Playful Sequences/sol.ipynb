```python
import random
import numpy as np
import torch
import os

def seed_everything(seed=20260322):
 random.seed(seed)
 os.environ['PYTHONHASHSEED'] = str(seed)
 np.random.seed(seed)

 try:
     torch.manual_seed(seed)
     torch.cuda.manual_seed(seed)
     torch.cuda.manual_seed_all(seed)
     torch.backends.cudnn.deterministic = True
     torch.backends.cudnn.benchmark = False
 except NameError:
     pass

seed_everything(20260322)
```

```python
import pandas as pd

df_train = pd.read_csv("train_data.csv")
df_test = pd.read_csv("test_data.csv")

df_train.head()
```

```python
df_train1 = df_train[df_train.subtaskID == 1]
df_train2 = df_train[df_train.subtaskID == 2]

df_test1 = df_test[df_test.subtaskID == 1]
df_test2 = df_test[df_test.subtaskID == 2]

df_train1.head()
```

```python
df_test.columns
```

```python
features = [col for col in df_test.columns if col != 'datapointID' and col != 'subtaskID']
target = 'Target'

X1 = df_train1[features]
X2 = df_train2[features]

y1 = df_train1[target]
y2 = df_train2[target]

X1_final = df_test1[features]
X2_final = df_test2[features]

from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
X1 = pd.DataFrame(imputer.fit_transform(X1), columns=features)
X1_final = pd.DataFrame(imputer.transform(X1_final), columns=features)

imputer = SimpleImputer(strategy='mean')
X2 = pd.DataFrame(imputer.fit_transform(X2), columns=features)
X2_final = pd.DataFrame(imputer.transform(X2_final), columns=features)

X1.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from lightgbm import LGBMClassifier
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import StackingClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score

estimators = [
    ('gb', LGBMClassifier(verbose=0)),
    ('svm', make_pipeline(StandardScaler(), LinearSVC()))]
model2 = StackingClassifier(estimators=estimators, final_estimator=LinearSVC())

model1 = LGBMClassifier(verbose=-1)
#model2 = LGBMClassifier(verbose=0)

print('Subtask 1: ', cross_val_score(model1, X1, y1, cv=3, scoring='f1_macro').mean())
print('Subtask 2: ',cross_val_score(model2, X2, y2, cv=3, scoring='f1_macro').mean())
```

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform

distributions = {
    "num_leaves": [30, 40, 50, 60, 70, 80, 100],
    "max_depth": [2, 3, 4, 5, 6, 7, 8],
    "learning_rate": uniform(0.01, 0.1),
    "n_estimators": [50, 100, 200, 300, 400, 500, 750],
    "min_child_samples": [10, 20, 30, 40, 50, 60, 70]
}
clf = RandomizedSearchCV(model1, distributions)
search = clf.fit(X1, y1)
search.best_params_
```

```python
model1 = LGBMClassifier(verbose=-1, **search.best_params_)
#model2 = LGBMClassifier(verbose=0)

print('Subtask 1: ', cross_val_score(model1, X1, y1, cv=3, scoring='f1_macro').mean())
print('Subtask 2: ',cross_val_score(model2, X2, y2, cv=3, scoring='f1_macro').mean())
```

```python
model1.fit(X1, y1)
preds_1 = model1.predict(X1_final)

model2.fit(X2, y2)
preds_2 = model2.predict(X2_final)
```

```python
df_1 = pd.DataFrame({
    "subtaskID": 1,
    "datapointID": df_test1.datapointID,
    "answer": preds_1
})

df_2 = pd.DataFrame({
    "subtaskID": 2,
    "datapointID": df_test2.datapointID,
    "answer": preds_2
})

df_1.head()
```

```python
submission = pd.concat([df_1, df_2])
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

```python
import random
import numpy as np
import torch

def set_seed(seed=42):
 random.seed(seed)
 np.random.seed(seed)
 torch.manual_seed(seed)
 if torch.cuda.is_available():
     torch.cuda.manual_seed_all(seed)
     torch.backends.cudnn.deterministic = True
     torch.backends.cudnn.benchmark = False

set_seed(42)
```

```python
import numpy as np
import os

root = 'dataset'

relic_ids = []
intensities = []
rotations = []
for relic in os.listdir(root):
    idx = relic.split('_')[1]
    relic_ids.append(idx)
    
    inten = np.load(f'dataset/{relic}/intensities.npy')
    intensities.append(inten)
    rot = np.load(f'dataset/{relic}/rotations.npy')
    rotations.append(rot)

intensities = np.array(intensities)
rotations = np.array(rotations)
```

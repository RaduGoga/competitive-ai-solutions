```python
import pandas as pd

df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

df_train.head()
```

```python
#subtasks

def get_cardio_risk(row):
    score = (row.Hypertension == 1) + (row.Diabetes == 1) + (row.BMI > 30)
    return score

def get_life_risk(row):
    score = (row.Smoking == 1) + (row.AlcoholConsumption > 2) + (row.PhysicalActivity < 1)
    return score

subtask1 = df_test.apply(get_cardio_risk, axis=1)
subtask2 = df_test.apply(get_life_risk, axis=1)

df_1 = pd.DataFrame({
    "PatientID": df_test.PatientID,
    "subtaskID": "Task1",
    "Answer": subtask1
})

df_2 = pd.DataFrame({
    "PatientID": df_test.PatientID,
    "subtaskID": "Task2",
    "Answer": subtask2
})

subtasks = pd.concat([df_1, df_2])
subtasks
```

```python
#preprocessing

df_train = df_train.drop(columns = ['DoctorInCharge', 'EyeColor'])
df_test = df_test.drop(columns = ['DoctorInCharge', 'EyeColor'])

df_train.head()
```

```python
features = [col for col in df_test.columns if col != 'PatientID']
target = 'Diagnosis'

X = df_train[features]
y = df_train[target]

X_final = df_test[features]

X.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from sklearn.model_selection import cross_val_score

model = RandomForestClassifier(random_state=42)

cross_val_score(model, X, y, cv=5, scoring='roc_auc')
```

```python
model.fit(X, y)
y_preds = model.predict_proba(X_final)[:, 1]
```

```python
df_preds = pd.DataFrame({
    "PatientID": df_test.PatientID,
    "subtaskID": "Task3",
    "Answer": y_preds
})

df_preds.head()
```

```python
submission = pd.concat([subtasks, df_preds])
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("muzee.csv")

df.head()
```

```python
#subtaskurile
subtask1 = len(df)
subtask2 = len(df[df["județul"] == "București"])
subtask3 = (df.isna().sum() >= 1).sum()
subtask4 = df['anul înființării'].value_counts().index[0]

df_1 = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "subtaskID": [1, 2, 3, 4],
    "answer": [subtask1, subtask2, subtask3, subtask4]
})

df_1
```

```python
subtask5 = df["județul"].value_counts()

df_2 = pd.DataFrame({
    "id": subtask5.index,
    "subtaskID": 5,
    "answer": subtask5.values
})

df_2.head()
```

```python
def scor_completitudine(row):
    nr_na = row.isna().sum()
    nr_bune = (len(df.columns) - nr_na)
    nr_total = len(df.columns)
    scor = nr_bune / nr_total * 100
    return scor

subtask6 = df.apply(scor_completitudine, axis=1)

df_3 = pd.DataFrame({
    "id": df._id,
    "subtaskID": 6,
    "answer": subtask6
})

df_3.head()
```

```python
subtask7 = subtask6.mean()

nr_cu_com_max = len(subtask6[subtask6 == subtask6.max()])
subtask8 = nr_cu_com_max / len(df) * 100

df_4 = pd.DataFrame({
    "id": [7, 8],
    "subtaskID": [7, 8],
    "answer": [subtask7, subtask8]
})

df_4
```

```python
submission = pd.concat([df_1, df_2, df_3, df_4])
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

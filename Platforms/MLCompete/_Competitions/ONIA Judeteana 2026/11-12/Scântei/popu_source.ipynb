```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
cl_res = []
th = 0.9
k_model = KMeans(5, random_state=42)

df = pd.DataFrame([], columns=['X', 'Y', 'seconds', 'train', 'id', 'res', 'cl'])
for i in data['Timestamp'].unique():
    part = data[data['Timestamp'] == i].copy()
    second = part['seconds'].unique()[0]
    k_model.fit(part[['X', 'Y']])
    centroids = k_model.cluster_centers_
    s = silhouette_score(part[['X', 'Y']], k_model.predict(part[['X', 'Y']]))
    print(i)
    print(s)
    
    part_res = [[f'{i}|{k}', f'{centroids[k][0]}|{centroids[k][1]}'] for k in range(5)]
    idk = pd.DataFrame([[centroids[i][0], centroids[i][1], second, 0, part_res[i][0], part_res[i][1], i] for i in range(5)], columns=['X', 'Y', 'seconds', 'train', 'id', 'res', 'cl'])
    if s > th:
        idk['train'] = 1
    df = pd.concat([df, idk])
    cl_res.extend(part_res)

len(cl_res)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
train_X, train_Y = df.loc[df['train'] == 1, ['seconds', 'X', 'cl']].copy(), df.loc[df['train'] == 1, ['seconds', 'Y', 'cl']].copy()
```

```python
lx = LinearRegression().fit(train_X[['seconds']], train_X['X'])
ly = LinearRegression().fit(train_Y[['seconds']], train_Y['Y'])
```

```python
df['predX'] = lx.predict(df[['seconds']])
df['predY'] = ly.predict(df[['seconds']])
```

```python
for_r4 = []
for i in df['cl'].unique():
    print(i)
    tx, ty = train_X[train_X['cl'] == i].copy(), train_Y[train_Y['cl'] == i].copy()
    lx.fit(tx[['seconds']], tx['X'])
    ly.fit(ty[['seconds']], ty['Y'])

    df.loc[df['cl'] == i, 'predX'] = lx.predict(df.loc[df['cl'] == i, ['seconds']])
    df.loc[df['cl'] == i, 'predY'] = ly.predict(df.loc[df['cl'] == i, ['seconds']])

    for_r4.append(f'{lx.predict([[500]])[0]}|{ly.predict([[500]])[0]}')
```

```python
def find_id(row):
    time = row['Timestamp']
    idd = row['num_id'] % 5
    return f'{time}|{idd}'

def find_coords(row):
    x = row['X']
    y = row['Y']
    return f'{x}|{y}'
```

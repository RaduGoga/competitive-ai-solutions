```python
import pandas as pd

df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

df_train.head()
```

```python
subtask1 = df_train.arena_type.nunique()
subtask2 = df_train.avg_speed_mps.max()
subtask3 = df_train.arena_type.mode()[0]
subtask4 = df_train.items_collected.max()

df_subtasks = pd.DataFrame({
    "robot_id": "GLOBAL",
    "subtaskID": [1, 2, 3, 4],
    "answer": [subtask1, subtask2, subtask3, subtask4]
})

df_subtasks
```

```python
dict1 = {'urban': 0, 'forest': 1, 'desert': 2, 'ice': 3}
dict2 = {'clear': 0, 'fog': 1, 'windy': 2}

df_train.arena_type = df_train.arena_type.map(dict1)
df_test.arena_type = df_test.arena_type.map(dict1)

df_train.weather = df_train.weather.map(dict2)
df_test.weather = df_test.weather.map(dict2)

df_train.head()
```

```python
features = [col for col in df_test.columns if col != 'robot_id']
target = 'strategy_label'

X = df_train[features]
y = df_train[target]

X_final = df_test[features]

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = pd.DataFrame(scaler.fit_transform(X), columns=features)
X_final = pd.DataFrame(scaler.transform(X_final), columns=features)

X.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

model = LogisticRegression(max_iter=10000000000)

cross_val_score(model, X, y, cv=5, scoring='f1_macro').mean()
```

```python
model.fit(X, y)
y_preds = model.predict(X_final)
```

```python
df_preds = pd.DataFrame({
    "robot_id": df_test.robot_id,
    "subtaskID": 5,
    "answer": y_preds
})

df_preds.head()
```

```python
submission = pd.concat([df_subtasks, df_preds])
submission
```

```python
submission.to_csv("submission.csv", index=False)
```

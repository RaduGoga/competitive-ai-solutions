```python
import pandas as pd
import numpy as np

df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

df_train.head()
```

```python
X = df_train.Text
X_final = df_test.Text

X.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vect = TfidfVectorizer(analyzer='char', ngram_range=((3, 5)))
X = vect.fit_transform(X)
X_final = vect.transform(X_final)

X
```

```python
from sklearn.metrics.pairwise import cosine_similarity
sims = cosine_similarity(X_final, X)
matches = np.argmax(sims, axis=1)

matches[:5]
```

```python
preds = []

for idx in matches:
    good = df_train.iloc[idx]["RevisedText"]
    preds.append(good)

preds[:5]
```

```python
submission = pd.DataFrame({
    "DatapointID": df_test.SampleID,
    "RevisedText": preds
})

submission
```

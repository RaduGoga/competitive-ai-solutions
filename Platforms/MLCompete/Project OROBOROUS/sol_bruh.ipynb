```python
import pandas as pd

df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

df_train.head()
```

```python
def is_prime(x):
    if x <= 1:
        return False
    if x == 2:
        return True
    if x % 2 == 0:
        return False
    for d in range(2, x//2):
        if x%d==0:
            return False
    return True

is_prime(11)
```

```python
preds = df_test.resonance_str.str.len().apply(is_prime)
preds
```

```python
preds.value_counts()
```

```python
submission = pd.DataFrame({
    "subtaskID": 1,
    "datapointID": df_test.id_hex,
    "answer": [int(pred) for pred in preds]
})

submission
```

```python
submission.to_csv("submission.csv", index=False)
```

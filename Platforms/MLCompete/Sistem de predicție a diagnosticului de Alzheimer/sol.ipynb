```python
import pandas as pd

df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

df_train.head()
```

```python
#datasetu asta e strict numeric si nici nare nicio valoare lips
#si nici nu prea sunt skewed datele
```

```python
#subtasks
subtask1 = []

for age in df_test.Age:
    subtask1.append(len(df_train[df_train.Age == age]))

subtask2 = []

for age in df_test.Age:
    nr1 = len(df_train[(df_train.Age == age) & (df_train.Smoking == 1)])
    nr2 = len(df_train[df_train.Age == age])
    subtask2.append(nr1/nr2*100)

df_1 = pd.DataFrame({
    "subtaskID": 1,
    "datapointID": df_test.PatientID,
    "answer": subtask1
})

df_2 = pd.DataFrame({
    "subtaskID": 2,
    "datapointID": df_test.PatientID,
    "answer": subtask2
})

subtasks = pd.concat([df_1, df_2])
subtasks
```

```python
df_train.columns
```

```python
features = ['Age', 'Gender', 'Ethnicity', 'EducationLevel', 'BMI', 'Smoking', 'AlcoholConsumption', 'PhysicalActivity', 'DietQuality', 'SleepQuality', 'FamilyHistoryAlzheimers', 'CardiovascularDisease',
        'Diabetes', 'Depression', 'HeadInjury', 'Hypertension', 'SystolicBP', 'DiastolicBP', 'CholesterolTotal', 'CholesterolLDL', 'CholesterolHDL',
        'CholesterolTriglycerides', 'MMSE', 'FunctionalAssessment', 'MemoryComplaints', 'BehavioralProblems', 'ADL', 'Confusion',
        'Disorientation', 'PersonalityChanges', 'DifficultyCompletingTasks','Forgetfulness']
target = 'Diagnosis'

X = df_train[features]
y = df_train[target]
X_final = df_test[features]

X.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

model = RandomForestClassifier(random_state=42)

cross_val_score(model, X, y, cv=5, scoring='roc_auc')
```

```python
model.fit(X, y) #prezicem proba pt ca e cu roc auc
y_preds = model.predict_proba(X_final)
y_preds[:10]
```

```python
df_preds = pd.DataFrame({
    "subtaskID": 3,
    "datapointID": df_test.PatientID,
    "answer": y_preds[:, 1]
})

df_preds.head()
```

```python
submission = pd.concat([subtasks, df_preds])

submission
```

```python
submission.to_csv("submission.csv", index=False)
```
